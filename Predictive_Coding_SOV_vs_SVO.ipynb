# Capstone Project: Predictive Coding and Syntactic Bottlenecks (SVO vs. SOV)
**Simulating LLM Surprisal and Cognitive Load in Cross-Linguistic Naturalistic Reading**

## 1. Project Objective & Hypothesis (From SOP)
The paradigm of language processing is shifting toward real-time tracking of naturalistic narratives. However, the Anglocentric focus on SVO (Subject-Verb-Object) structures creates a critical blind spot regarding the 'syntactic cognitive load' in SOV languages (like Korean). 

**Hypothesis:** The head-final structure of SOV languages forces a **'delayed semantic commitment'**, expanding the working memory buffer. Consequently, when the verb finally appears at the end of the sentence, it triggers a massive resolution of syntactic surprisal, which will be reflected in a sharp spike in cognitive load (simulated here as LLM Surprisal / N400 amplitude) compared to the more evenly distributed load in SVO languages.

## 2. Technical Stack
* **Data Simulation:** `pandas`, `numpy` (Modeling word-by-word reading trajectories)
* **Visualization:** `seaborn`, `matplotlib` (Plotting cognitive load spikes)

## 3. References & Theoretical Foundations
* **Gibson, E. (1998).** Linguistic complexity: locality of syntactic dependencies. *Cognition*, 68(1), 1-76. *(Theoretical basis for Dependency Locality Theory and working memory load in head-final SOV structures)*
* **Kwon, N., Polinsky, M., & Kluender, R. (2013).** Subject/object processing asymmetries in Korean relative clauses: Evidence from ERP data. *Language and Cognitive Processes*, 28(10), 1379-1408. *(Empirical ERP evidence of syntactic processing bottlenecks and P600/N400 delays in Korean)*
* **Levy, R. (2008).** Expectation-based syntactic comprehension. *Cognition*, 106(3), 1126-1177. *(Mathematical foundation for quantifying syntactic surprisal as logarithmic predictive processing error: $Surprisal = -\log P(w_t \mid context)$)*
* **Huth, A. G., de Heer, W. A., Griffiths, T. L., Theunissen, F. E., & Gallant, J. L. (2016).** Natural speech reveals the semantic maps that tile human cerebral cortex. *Nature*, 532(7600), 453-458. *(Methodological anchor for naturalistic narrative processing across cerebral cortex)*

## 4. Methodological Note: A Priori Computational Simulation

It is explicitly noted that the data visualised in this repository is **synthetically simulated (a priori generative modeling)** rather than empirically collected neuroimaging data. 

In modern computational neuroscience and pre-registered research, simulating expected neurophysiological trajectories before data acquisition is standard practice. The primary objectives of this simulation are:
1. **Mathematical Formalization:** To translate a theoretical linguistic hypothesis (delayed semantic commitment in SOV structures) into an explicit, quantitative predictive model.
2. **Pipeline Feasibility:** To demonstrate programmatic readiness to execute time-series visualizations and statistical models immediately upon gaining access to real-world MEG/fMRI datasets at Macquarie University.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Simulate Word-by-Word Cognitive Load (Surprisal / N400)
# English (SVO): The boy (1) / ate (2) / the delicious (3) / apple (4).
# Korean (SOV): 그 소년은 (1) / 맛있는 (2) / 사과를 (3) / 먹었다 (4).

data = {
    'Word_Position': [1, 2, 3, 4, 1, 2, 3, 4],
    'Language_Type': ['English (SVO)', 'English (SVO)', 'English (SVO)', 'English (SVO)', 
                      'Korean (SOV)', 'Korean (SOV)', 'Korean (SOV)', 'Korean (SOV)'],
    # SVO: Load peaks slightly at verb, then resolves.
    # SOV: Memory buffer builds up, massive spike at the final verb (Delayed Commitment).
    'Simulated_Surprisal': [2.1, 4.5, 3.2, 2.8, 
                            2.5, 3.8, 4.2, 7.9] 
}

df_sim = pd.DataFrame(data)

# 2. Visualize the Syntactic Bottleneck (Choking Point)
plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")

# Line plot to show the trajectory of cognitive load
sns.lineplot(
    data=df_sim, 
    x='Word_Position', 
    y='Simulated_Surprisal', 
    hue='Language_Type',
    marker='o', 
    linewidth=3, 
    markersize=10,
    palette=['#1f77b4', '#d62728'] # Blue for SVO, Red for SOV
)

# Annotations for theoretical concepts
plt.axvline(x=4, color='black', linestyle='--', alpha=0.5)
plt.text(3.5, 7.0, 'Delayed Semantic\nCommitment (SOV Verb)', fontsize=11, color='#d62728', fontweight='bold')
plt.text(1.8, 5.0, 'Early Resolution\n(SVO Verb)', fontsize=11, color='#1f77b4', fontweight='bold')

plt.title("Simulated Cognitive Load Trajectory: SVO vs. SOV Structures", fontsize=16, fontweight='bold')
plt.xlabel("Sentence Position (Time Course)", fontsize=13)
plt.ylabel("Cognitive Load (Simulated LLM Surprisal / N400)", fontsize=13)
plt.xticks([1, 2, 3, 4], ['Subject / Noun 1', 'Verb (SVO) / Adj (SOV)', 'Adj (SVO) / Object (SOV)', 'Object (SVO) / Verb (SOV)'])
plt.legend(title='Syntactic Structure')

plt.tight_layout()
plt.show()
```
